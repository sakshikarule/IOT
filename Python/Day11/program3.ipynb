# Data Visualization

## Line Plot

```python
# import the requried packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the datset

df=pd.read_csv("Salary_Data1.csv")
df.head()
```

### plt.scatter(x-axis,y-axis,..)

```python
# plt.scatter(df["YearsExperience"],df["Salary"],label="Salary")
plt.plot(df["YearsExperience"],df["Salary"],label="Salary",color="green")
plt.plot(df["YearsExperience"],df["Expences"],label="Expense",color="red")

plt.scatter(df["YearsExperience"], df['Salary'], label="Salary", 
color="green")
plt.scatter(df["YearsExperience"], df['Expences'], 
label="Expences", color="red")

# add the chart information to make it more readable
plt.title("YearsExperience VS Salary")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
```

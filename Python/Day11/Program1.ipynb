# Create pandas using dictionary

```python
s1=pd.Series({"Name":"sakshi","org":"Sunbeam","id":123,"course":"python"})
print(s1)
print(f"s1[Name] = {s1['Name']}")
print(f"s1[0] = {s1[0]}")
```

```python
students=pd.Series(["s1","s2","s3","s4"])
marks=pd.Series([25,30,35,40])
print(students)
print(marks)
```

```python
students=pd.Series(["s1","s2","s3","s4"])
marks=pd.Series([25,30,35,40],index=students)
print(marks)
```

```python
marks=pd.Series([25,30,35,40])
students=pd.Series(["s1","s2","s3","s4"],index=marks)
print(students)
```

# Create Dataframe using Dictiionary

```python
def function2():
    p1=({'name':'p1','age':25,'id':1234})
    p2=({'name':'p2','age':35,'id':5678})
    p3=({'name':'p3','age':30,'id':3456})
    p4=({'name':'p4','age':40,'id':6789})

    s1=pd.Series(p1)
    s2=pd.Series(p2)
    s3=pd.Series(p3)
    s4=pd.Series(p4)

    df=pd.DataFrame([s1,s2,s3,s4])
    print(s1)
    print(s2)
    print(s3)
    print(s4)
    
    print(df)
function2()
```

```python
def function3():
    p1=({'name':'p1','age':25,'id':1234})
    p2=({'name':'p2','age':35,'id':5678})
    p3=({'name':'p3','age':30,'id':3456})
    p4=({'name':'p4','age':40,'marks':35})

    s1=pd.Series(p1)
    s2=pd.Series(p2)
    s3=pd.Series(p3)
    s4=pd.Series(p4)

    df=pd.DataFrame([s1,s2,s3,s4])
    print(df)
function3()
```

```python
def function4():
    # filtering
    salaries=pd.Series([30,35,45,40,70,90,80,75])
    # find salaries above 70k
    print(f"Salaries above 70k = \n{salaries[salaries>70]}")
function4()
```

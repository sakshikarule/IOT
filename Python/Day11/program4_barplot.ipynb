## Data visualization

### bar chart

```python
# import the required packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data
fruits=['apple','blueberry','cherry','oranges','Gava']
count=[40,20,30,100,50]
bar_colors= ['tab:blue','tab:green','tab:red','tab:orange']
```

## EDA(Exploratory data Analysis)

```python
plt.bar(fruits,count,label=fruits,color=bar_colors)

plt.title("Fruit supply information")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.grid(alpha=0.2)
plt.legend()
```

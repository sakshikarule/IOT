## Pandas

```python
#import the requrired packages
import numpy as np
import pandas as pd
```

```python
#prerequisties

import warnings
warnings.filterwarnings(action="ignore")
```

```python
a1=[10,20,30,40,50]
print(f"A1 ={a1} Type={type(a1)}")
```

```python
n_array=np.array([10,20,30,40,50])
print(f"N_ARR ={n_array}  Type={type(n_array)}")
```

```python
s1 =pd.Series([10,20,30,40,50])
print(f"Series =\n{s1}     Type={type(s1)}")
```

```python
def function1():
    n_series=pd.Series([10,20,30,40,50])
    print(f"Type={type(n_series)}")
    print(f"DType={type(n_series.dtype)}")
    print(f"Size={type(n_series.size)}")
    print(f"DIM={type(n_series.ndim)}")
    print(f"N_Bytes={type(n_series.nbytes)}")

function1()    
```

```python
n_series=pd.Series([10,20,30,40,50,60])
print(f"Keys ={n_series.keys()}")
print(f"Value{n_series.values}")
```
